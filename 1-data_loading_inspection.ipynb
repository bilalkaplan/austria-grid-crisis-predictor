```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. Data Loading and Extracting the Relevant Feature
# Make sure you typed the name of the unzipped file correctly!
df = pd.read_csv('time_series_60min_singleindex.csv', parse_dates=['utc_timestamp'], index_col='utc_timestamp')

# Just like the "Close" price in the plan, we only extract the column we will predict
solar_data = df[['AT_solar_generation_actual']].dropna()
print(f"Total Hourly Data Count: {len(solar_data)}")

# 2. Data Normalization (Scaling)
# According to the guide, we use MinMaxScaler to bring the data into the [-1, 1] range
scaler = MinMaxScaler(feature_range=(-1, 1))
solar_data_scaled = scaler.fit_transform(solar_data.values)

print("\nFirst 5 normalized (scaled) data points:")
print(solar_data_scaled[:5])
```
